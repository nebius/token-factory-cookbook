# Replaying the mobile repair benchmark

Run this notebook from the recipe directory in a Python 3.12+ kernel with the browser dependencies described in README.md. The saved predictions let you reproduce the evaluation without a GPU or API key. For fresh training, use the JupyterLab Terminal walkthrough in the README.

```python
import json, os, subprocess, sys
from pathlib import Path
from IPython.display import Image, display
env = os.environ.copy()
env["PATH"] = str(Path(sys.executable).parent) + os.pathsep + env["PATH"]
subprocess.run(["bash", "replay.sh"], env=env, check=True)
out = Path("replay-results")
scores = json.loads((out / "scores.json").read_text())
for split, result in scores.items():
    print(split, {name: values["pass"] for name, values in result["arms"].items()})
display(Image(filename=str(out / "before-after.png"), width=900))
```

Open `replay-results/showcase.html` to inspect all 32 pages, the CSS patches and the five viewport sizes.
