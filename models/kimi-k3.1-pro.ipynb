## Set up the client

Use an environment variable or enter your API key at the hidden prompt. The key is not printed.


```python
import os
from getpass import getpass
from openai import OpenAI

api_key = os.environ.get("NEBIUS_API_KEY") or getpass("Token Factory API key: ")
client = OpenAI(
    api_key=api_key,
    base_url="https://api.tokenfactory.nebius.com/v1/",
)
```

## Make your first request

Expect a retry function and explanation; the exact answer varies. Review and test generated code before using it.


```python
response = client.chat.completions.create(
    model="moonshotai/Kimi-K3.1-Pro",
    messages=[{
        "role": "user",
        "content": "Write a Python function that retries a request up to three times with exponential backoff.",
    }],
)

print(response.choices[0].message.content)
```

## Stream an answer

This displays answer text as it arrives. The model may take time before producing answer text.


```python
stream = client.chat.completions.create(
    model="moonshotai/Kimi-K3.1-Pro",
    messages=[{"role": "user", "content": "Explain exponential backoff in three bullet points."}],
    stream=True,
)

for chunk in stream:
    if chunk.choices:
        text = chunk.choices[0].delta.content
        if text:
            print(text, end="", flush=True)
print()
```
