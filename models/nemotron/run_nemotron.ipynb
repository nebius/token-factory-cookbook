# Running NVIDIA Nemotron Models on Nebius Token Factory

Minimal local example. Works with any Nemotron model on Token Factory. Set `MODEL_NAME` below to the one you want to use:

| Model ID | Parameters | Context |
|----------|------------|---------|
| `nvidia/Nemotron-3_5-Lightning` | 30B total / 3B active | 1M |
| `nvidia/nemotron-3-super-120b-a12b` | 120B total / 12B active | 256K |
| `nvidia/Nemotron-3-Ultra-550b-a55b` | 550B total / 55B active | 1M |

See the full list at [tokenfactory.nebius.com/models](https://tokenfactory.nebius.com/models).

## Setup:

```bash
uv sync
cp env.example .env   # then paste your NEBIUS_API_KEY into .env
uv run jupyter lab run_nemotron.ipynb
```

For a Google Colab version, see [run_nemotron_colab.ipynb](run_nemotron_colab.ipynb).

## Run


````python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

client = OpenAI(
    base_url="https://api.tokenfactory.nebius.com/v1/",
    api_key=os.environ["NEBIUS_API_KEY"],
)

# Pick any Nemotron model available on Token Factory
MODEL_NAME = "nvidia/Nemotron-3_5-Lightning"
# MODEL_NAME = "nvidia/nemotron-3-super-120b-a12b"
# MODEL_NAME = "nvidia/Nemotron-3-Ultra-550b-a55b"
````

````python
completion = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[{"role": "user", "content": "What is the capital of France?"}],
)

print(completion.choices[0].message.content)
````
