# How to use Inkling 2 Small

Make a request through Nebius Token Factory using `thinkingmachines/Inkling-2-Small`.

[Model details](https://tokenfactory.nebius.com/?modals=endpoint-details&model-id=thinkingmachines/Inkling-2-Small) · [Full quickstart](https://github.com/nebius/token-factory-cookbook/blob/main/models/inkling-2-small.md)

You need a Token Factory API key and project access to the model. Requests use your project balance. This notebook does not contain recorded model output.

## Set up the client

Use an environment variable or enter your API key at the hidden prompt. The key is not printed.

```python
import os
from getpass import getpass
from openai import OpenAI

api_key = os.environ.get("NEBIUS_API_KEY") or getpass("Token Factory API key: ")
client = OpenAI(
    api_key=api_key,
    base_url="https://api.tokenfactory.nebius.com/v1/",
)
```

## Make your first request

Expect operating income of $21.6M before and $22.5M after, a rise of $0.9M; the wording varies. The model reasons before it answers, and the usage shows how many tokens that took.

```python
response = client.chat.completions.create(
    model="thinkingmachines/Inkling-2-Small",
    messages=[{
        "role": "user",
        "content": (
            "Revenue grew from $120M to $150M while operating margin fell from 18% to 15%. "
            "Did operating income rise or fall, and by how much? Show the arithmetic."
        ),
    }],
)

print(response.choices[0].message.content)
print("Reasoning tokens:", response.usage.completion_tokens_details.reasoning_tokens)
```

## Stream an answer

This displays answer text as it arrives. The model may take time before producing answer text.

```python
stream = client.chat.completions.create(
    model="thinkingmachines/Inkling-2-Small",
    messages=[{"role": "user", "content": "Explain gross margin and operating margin in three bullet points."}],
    stream=True,
)

for chunk in stream:
    if chunk.choices:
        text = chunk.choices[0].delta.content
        if text:
            print(text, end="", flush=True)
print()
```
